```python
#Closure
def counter():
    count = 0
    def increment():
        nonlocal count
        count += 1
        return count
    return increment

c = counter()
print(c())  # 1
print(c())  # 2
print(c())  # 3
```

```python
person_info = "Name: John Doe 22222, Age: 30, Email: john.doe@example.com, Phone: 9873464626"
re.search(r"\d+",person_info)
print(re.search(r"\d{10}",person_info))
print(re.search(r"\d{2,5}",person_info))
print(re.search(r"[\w+\.-]+@[\w+\.-]+\.\w+", person_info))
```

```python
stri="COntact: +1 292 555 1234"
str2="COntact: +1 (292) 555 1234"
str3="COntact: 292-555-1234"

print(re.search(r"\+\d \d{3} \d{3}\s\d{4}",stri))
print(re.search(r"(\+1)?\d{3} \d{3}\s\d{4}",stri))
print(re.search(r"(\+1)?\(?\d{3}\)? \d{3}\s\d{4}",str2))    # +1 and () are optional so can work without
print(re.search(r"(\+1)?\(?\d{3}\)?[ -]\d{3}[ -]\d{4}",str3) )
```

```python
stri=["COntact: +1 292 555 1234","COntact: +1 (292) 555 1234",
"COntact: +1 (292) 555 1204",
"COntact: 292-555-1294",]
for st in stri:
  matc =re.search(r"(\+1)?\(?\d{3}\)?[ -]\d{3}[ -]\d{4}",st)    #[ -] means it can take either of the two cases
  print(matc)
```

#INHERITANCE

```python
class Person:
  def __init__(self,name,age):
    self.name=name
    self.age=age

  def get_name(self):
    return self.name

class Manager(Person):
  def __init__(self,name,age,salary):
    super().__init__(name,age)
    self.salary=salary

  def info(self):
    print(self.name,self.age,self.salary)

class Employee(Person):
  def __init__(self,name,age,desk_num):
    super().__init__(name,age)
    self.desk_num=desk_num

  def info(self):
    print(self.name,self.age,self.desk_num)
```

```python
man=Manager("Ram",11,30000000)
man.info()
man.get_name()
```

```python
em=Manager("Ram",11,30000000)
em.info()
em.get_name()
```
