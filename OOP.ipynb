```python

class Point:

  def __init__(self,x,y):
    self.x=x
    self.y=y
    self.dist=(self.x**2 + self.y**2)**0.5

  def __str__(self):
    return f"({self.x},{self.y})"

  def __repr__(self):
    return f"X-Coord = {self.x}, Y-Coord = {self.y}"

  def __add__(self, other):                         #Returns a new class after adding
    return Point(self.x+other.x,self.y+other.y)

  def __iadd__(self,other):                         #Returns same class
    self.x += other.x
    self.y += other.y
    return self

  def __mul__(self, scalar):
    return Point(self.x*scalar,self.y*scalar)

  def __eq__(self, other):
    return (self.x==other.x and self.y==other.y)

  def __lt__(self, other):
    return  (self.dist < other.dist)

  def __gt__(self, other):
    return (self.dist > other.dist)

  def distance_from_origin(self):
    return self.dist

  def __hash__(self):
    return hash((self.x,self.y))    #hash() converts an object(immutable) into a number so Python can store and find it quickly in dictionaries and sets


#DRY (Do not Repeat Yourself)
```

```python
a=Point(3,4)
b=Point(4,7)
c=a+b
d=a*2
print(a==b,c.dist,d.dist)
print(a>b,a<b,a.distance_from_origin())
c+=b
print(c.x)
se={Point(1,2), Point(2,3), Point(1, 2)}
print(se)
```

```python
class Bank:

  total_balance=0

  def __init__(self,name,balance):
    self.name=name
    self.balance=balance
    Bank.total_balance+=balance

  def deposit(self,amount):
    self.balance+=amount
    Bank.total_balance+=amount

  def withdraw(self,amount):
    if self.balance>=amount:
      self.balance-=amount
      Bank.total_balance-=amount

    elif Bank.total_balance<amount:
      print("Insufficient Bank Funds")

    else:
      print("Insufficient Personal Funds")


  def check_balance(self):
    print(f"Personal Balance: {self.balance}")


```
