```python
# A Python dictionary using various data types
from datetime import datetime as dt
import datetime as dt
data = {
    "names": ["Alice", "Bob", "Charlie", "Diana"],
    "dates": ["2026-01-01", "2026-01-05", "2026-01-08", "2026-01-09"],
    "scores": [95, 88, 92, 100]
}
```

```python
with open("data.json","w") as f:
  json.dump(data,f) # Ensure data.json has content

with open("data.json","r") as f:
  b=json.load(f)

print("Date","Name","Score", sep=" | ",end="\n\n")
for i in range(len(b["names"])):
    if b["scores"][i]>=90:
      print(dt.datetime.strftime(dt.datetime.strptime(b["dates"][i],"%Y-%m-%d"),"%d %B, %Y"),b["names"][i],b["scores"][i], sep=" | ")
      print("-"*100)
```
