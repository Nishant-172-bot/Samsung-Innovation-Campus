```python
x=1

def func():
  x=2
  def inner():
    nonlocal x      #Only to update values, you can access values directly to use
    x += 1
    print(x)
  print(x)
  inner()
  print(x)

del x   #remove global
func()
```

##Build a counter that increases everytime it is called

```python
def counter(count=1):
  def inner():
    nonlocal count
    count+=1
    print(f"Cuurent Count : {count}")
  print(f"Initialized count as : {count}")
  return inner            #return inner cause it doesnt need any args

my_counter=counter(10)    #initialize outer only once
my_counter()
my_counter()
my_counter()
my_counter2=counter(22)
my_counter2()
```

##map, reduce, filter, lambda

```python
a=[1,2,3,4,5,6]
[i**2 for i in a]
```

```python
b=lambda x:x**2
list(map(b, a))         #applies an operation to each element of an iterable
```

```python
b=lambda x:x**2
sum(map(b,a))
```

```python
s_list = ['abc', 'bcd', 'bcdefg', 'abba', 'cddc', 'opq']
list(map( len , s_list ))     #dont use lambda for just single operations
```

```python
list(filter(lambda x:len(x)>3,s_list))      #filters the elements who yield true to a condition
```

```python
a=[1,2,3,4,5,6]
list(filter(lambda x: not x % 2,a))
```

```python
c=[1,23,4,[],"",None]
list(filter(None,c))
```

```python
data=[{"name":"Ram","Age":18},
      {"name":"Ram2","Age":22},
      {"name":"Ram3","Age":88},]
list(filter(lambda x : x["Age"]>20,data))
```

```python
from functools import reduce

nums = [1, 2, 3, 4]

result = reduce(lambda a, b: a + b, nums)   #applies a binary function cumulatively to each elemnt of an iterable -> (((1+2)+3)+4)
print(result)  # 10
```

```python
from functools import reduce
from operator import mul

nums = [1, 2, 3, 4]

result = reduce(mul, nums)          #(((1*2)*3)*4)
print(result)
```
