```python
# Bubble Sort Implementation
def bubble_sort(arr):
    n = len(arr)
    # Traverse through all array elements
    for i in range(n):
        # Last i elements are already in place
        for j in range(0, n - i - 1):
            # Traverse the array from 0 to n-i-1
            # Swap if the element found is greater than the next element
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

# Example usage for Bubble Sort
my_list = [64, 34, 25, 12, 22, 11, 90]
print("Original list for Bubble Sort:", my_list)
print("Sorted list for Bubble Sort:", bubble_sort(my_list.copy()))
```

```python
# Selection Sort Implementation
def selection_sort(arr):
    n = len(arr)
    # Traverse through all array elements
    for i in range(n):
        # Find the minimum element in remaining unsorted array
        min_idx = i
        for j in range(i + 1, n):
            if arr[min_idx] > arr[j]:
                min_idx = j

        # Swap the found minimum element with the first element
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

# Example usage for Selection Sort
my_list = [64, 25, 12, 22, 11]
print("Original list for Selection Sort:", my_list)
print("Sorted list for Selection Sort:", selection_sort(my_list.copy()))
```

```python
# Insertion Sort Implementation
def insertion_sort(arr):
    # Traverse through 1 to len(arr)
    for i in range(1, len(arr)):
        key = arr[i]
        # Move elements of arr[0..i-1], that are greater than key,
        # to one position ahead of their current position
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

# Example usage for Insertion Sort
my_list = [12, 11, 13, 5, 6]
print("Original list for Insertion Sort:", my_list)
print("Sorted list for Insertion Sort:", insertion_sort(my_list.copy()))
```

```python
# Quick Sort Implementation
def quick_sort(arr):
    # Base case: if the array has 0 or 1 element, it's already sorted
    if len(arr) <= 1:
        return arr
    else:
        # Choose the pivot element (here, the last element)
        pivot = arr[-1]
        # Partition the array into elements less than pivot and greater than pivot
        less = [x for x in arr[:-1] if x <= pivot]
        greater = [x for x in arr[:-1] if x > pivot]
        # Recursively sort the sub-arrays and combine them
        return quick_sort(less) + [pivot] + quick_sort(greater)

# Example usage for Quick Sort
my_list = [10, 7, 8, 9, 1, 5]
print("Original list for Quick Sort:", my_list)
print("Sorted list for Quick Sort:", quick_sort(my_list.copy()))
```

```python
# Merge Sort Implementation
def merge_sort(arr):
    # Base case: if the array has 0 or 1 element, it's already sorted
    if len(arr) <= 1:
        return arr

    # Find the middle point to divide the array into two halves
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    # Recursively sort the two halves
    left_half = merge_sort(left_half)
    right_half = merge_sort(right_half)

    # Merge the sorted halves
    return merge(left_half, right_half)

def merge(left, right):
    result = []
    i = 0  # Pointer for left array
    j = 0  # Pointer for right array

    # Compare elements from left and right halves and add the smaller one to result
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # Add any remaining elements from the left half
    while i < len(left):
        result.append(left[i])
        i += 1

    # Add any remaining elements from the right half
    while j < len(right):
        result.append(right[j])
        j += 1
    return result

# Example usage for Merge Sort
my_list = [38, 27, 43, 3, 9, 82, 10]
print("Original list for Merge Sort:", my_list)
print("Sorted list for Merge Sort:", merge_sort(my_list.copy()))
```
