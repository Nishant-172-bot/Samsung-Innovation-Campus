#Decorator?
Functions in Python are first-class objects, meaning:

You can pass them as arguments

Return them from other functions

Assign them to variables

This makes decorators possible.

```python
#A decorator is a function that modifies or enhances another function without changing its actual code.
```

```python
from functools import lru_cache

@lru_cache(maxsize=128)
def add(a, b):
    print("Computing...")
    return a + b

print(add(2, 3))  # Computes
print(add(2, 3))  # Returns cached result (no "Computing...")
```

```python
from functools import wraps
```

```python
@wraps    #to account for args and kwargs
def dec(func):
  def inner():
    #preprocessing
    print("Running complex preprocessing(Pre)")
    func()
    print("Running complex preprocessing(Post)")
    #post_processing
  return inner

#or

def dec2(func):
  def inner(*args,**kwargs):
    #preprocessing
    print("Running complex preprocessing(Pre)")
    val = func(*args,**kwargs)
    print("Running complex preprocessing(Post)")
    #post_processing
    return val
  return inner
```

```python
def test_function():
  print("I am the function")
```

```python
func=dec(test_function)
```

```python
type(func)
```

```python
test_function()
```

```python
func()
```

```python
"""
Same process as above
"""
@dec
def test_function2():
  print("I am the function")
```

```python
test_function2()
```

```python
@dec2
def add(a):
  return a*2
```

```python
add(3)
```

```python
@dec
def square2(x):
  return x ** 2
```

```python
square2(10)
```

```python
@wraps
def logger(func):
  def inner(*args,**kwargs):
    #preprocessing
    print(func,args,kwargs)
    val = func(*args,**kwargs)
    print(val)
    #post_processing
    return val
  return inner
```

```python
@logger
def square2(x):
  return x**2
```

```python
square2(10)
```

```python
import time
```

```python
def timer(func):
  def inner(*args,**kwargs):
    t1=time.time()
    val=func(*args,**kwargs)
    t2=time.time()
    print(f"time Taken {t2-t1}s")
    return val
  return inner
@timer
def sleepx(x):
  return f"Slept for {x}s"
sleepx(3)
```

```python
class TestClass:

  def __init__(self,x=10):
    self.temp_c=x

  @property
  def temp_f(self):
    #no call needed
    return self.temp_c*9/5 -32

  @staticmethod
  def temp_far(x):
    #self not used here
    some_value=10
    other=10
    return x*some_value + other

  @classmethod
  def classmeth(cls,x):
    some_val=(x-20)/10
    return cls(some_val)
```

```python
t=TestClass()
t.temp_f
```
