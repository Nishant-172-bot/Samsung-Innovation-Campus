#Tromino  Problem

```python
def create_board(n):
  return [[0 for _ in range(n)] for _ in range(n)]


def twobytwo(board, start_row, start_col):
  filled = 1
  for i in range(start_row, start_row + 2):
    for j in range(start_col, start_col + 2):
      if board[i][j] != 'f':
        board[i][j] = filled
  return board

def original(board, n, start_row=0, start_col=0):
  if n == 2:
    return twobytwo(board, start_row, start_col)
  half_n = n // 2
  original(board, half_n, start_row, start_col)
  original(board, half_n, start_row, start_col + half_n)
  original(board, half_n, start_row + half_n, start_col)
  original(board, half_n, start_row + half_n, start_col + half_n)

  return board
```

```python
board = create_board(8)

def fill(board,n):
  for i in range(0, n, 4):
    for j in range(0, n, 4):
      board[i][j] = 'f'
fill(board,8)
original(board, 8, 0, 0)

for r_idx, row in enumerate(board):
    for c_idx, val in enumerate(row):
        if val == 'f':
            board[r_idx][c_idx] = 0

board
```
