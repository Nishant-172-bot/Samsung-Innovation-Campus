#Recursion
-> Recursion relation
  -Input should be sammler once we go into the depths of the recusive chain
-> Stop condition
  -stop from going to infinite loop

```python
def factorial(n):
  return n*factorial(n-1) if n > 1 else 1

factorial(5)
```

```python
def print_numbers(n):
  if n >= 1:
    print(n)
    print_numbers(n-1)
  else:
    return 0

print_numbers(10)
```

#Class

```python
class Name:
  a=10
  b=20

  def __str__(self):
    return self.name
    #defines what happens when we print instance

  def __repr__(self):
    return f"Name = {self.name}"
    #repreesentation of class

  def __init__(self,name):
    self.name=name
    print("Initialized")

  def __len__(self):
    return len(self.name)

  def greet(self,name=""):
    print(f"Hello, {self.name} and {name}")
```
