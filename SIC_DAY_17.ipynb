```python
population_2020 = {
    "China": 1424929781,
    "India": 1380004385,
    "United States": 331002651,
    "Indonesia": 273523615,
    "Pakistan": 220892340,
    "Brazil": 212559417,
    "Nigeria": 206139589,
    "Bangladesh": 164689383,
    "Russia": 145934462,
    "Mexico": 128932753
}
population_2021 = {
    "China": 1425893465,
    "India": 1393409038,
    "United States": 332915073,
    "Indonesia": 276361783,
    "Pakistan": 225199937,
    "Brazil": 214326493,
    "Nigeria": 211400708,
    "Bangladesh": 166303498,
    "Russia": 145912025,
    "Mexico": 130262216
}
land_area_sq_km = {
    "China": 9424703,
    "India": 2973190,
    "United States": 9147590,
    "Indonesia": 1811570,
    "Pakistan": 770880,
    "Brazil": 8358140,
    "Nigeria": 910770,
    "Bangladesh": 130170,
    "Russia": 16376870,
    "Mexico": 1943950
}

```

```python
s_2020=pd.Series(population_2020)
s_2021=pd.Series(population_2021)
s_area=pd.Series(land_area_sq_km)
```

```python
print(s_2020)
s_2020.describe()
```

```python
print(s_2021)
s_2021.describe()
```

```python
print(s_area)
s_area.describe()
```

```python
growth=s_2021-s_2020
growth
```

```python
plt.barh(growth.index, growth.values)
plt.xlabel('Population Growth')
plt.ylabel('Country')
plt.title('Population Growth from 2020 to 2021')
# plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

```python
density=s_2020/s_area
density
```

```python
plt.bar(density.index, density.values)
plt.xlabel('Population Density')
plt.ylabel('Country')
plt.title('Population Density of 2020 ')
plt.show()
```
