#WAP to find sum and average of marks n students for n subjects

```python
amarks=[]
stu=int(input("No of students? "))
sub=int(input("No of subjects? "))
print()
for i in range(stu):
  smarks=[]
  sum=0;avg=0
  for j in range(sub):
    ma=int(input("Marks:"))
    smarks.append(ma)
    sum+=ma
  avg=sum/sub
  print(f"Sum: {sum} and Average: {avg} of student {i+1}\n")
  amarks.append(smarks)
print(amarks)
```

#WAP to find the first repeated element in the list

```python
arr=[32,"12",2,3,"1"]
flag=False
for i in range(len(arr)):
  if flag==True:
    break
  for j in range(i+1,len(arr)):
    if arr[i]==arr[j]:
      print(f"First repeated elem: {arr[i]}")
      flag=True
      break
    else:
      continue
if flag==False:
  print("No repeated elements")
```
