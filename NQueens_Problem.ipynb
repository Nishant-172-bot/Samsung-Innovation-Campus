```python
def create_board(N):
  return [-1]*N
```

```python
def is_valid(board,row):
  curr_value=board[row]
  if curr_value in board[:row]:   #check same column
      return False
  for i,k in enumerate(range(row,0,-1)):
      if (board[i]==(curr_value - k)) or (board[i]==(curr_value + k)):
          return False
  return True
```

```python
def queens(board, row=0, all_results=None):
  if all_results is None:
    all_results = []

  if row == len(board):
    all_results.append(board.copy())
    return

  for col in range(len(board)):
    board[row] = col
    if is_valid(board, row):
      queens(board, row + 1, all_results)
    board[row] = -1

  return all_results
```
