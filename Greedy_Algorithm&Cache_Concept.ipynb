```python
possible={1000:0,500:2,100:1,50:1,10:1,5:10,1:100}

def ch(change):
  global possible
  res={}
  temp=change

  for i,v in possible.items():
      if change >= i and v!=0:
        gr=change//i if v > change//i else v
        res[i]=gr
        change -= gr*i
        possible[i]=possible[i]-1
        continue

  if change!=0:
      return None

  return f"The change for {temp} => {res}"
```

```python
print(ch(1170))
print(ch(170))
```

```python
possible={1000:0,500:2,100:1,50:1,10:1,5:10,1:100}
j=0
def rec(change,possible,ind):
  res={}
  temp=change
  global j
  j+=1
  for i,v in possible.items():

      if change >= i and v!=0 and i<=ind:
        gr=change//i if v > change//i else v
        res[i]=gr
        change -= gr*i
        possible[i]=possible[i]-1
        continue

  x=[sum(list[res.values()]),rec(temp,possible,ind[j])]
  return x
```

```python
def ch(change):
  possible={1000:0,500:2,100:1,50:1,20:2,10:1,5:1,1:100}
  res1={}
  res2={}
  while change > 0:
    for i,v in possible.items():
      if change == (change//i)*i and v >= change//i:
        count=change//i
        res1[i]=count
        change-=count*i
        break

    for i,v in possible.items():
      if change >= i and v!=0:
        count=change//i if v > change//i else v
        res2[i]=count
        change -= count*i
        possible[i]=possible[i]-1
        break
  return res1 if res1.values()>res2.values() else res2
```

```python
ch(60)
```

```python
memo={}
def fib(n):
  f=None
  if n in memo:
    f="Found in memo"
    res = memo[n]
  if n<=1:
    return 1
  else:
    res=fib(n-1)+fib(n-2)
  memo[n]=res
  return res if not f else f"found {res} in memo"
```

```python
for i in range(10):
  fib(i)
```
