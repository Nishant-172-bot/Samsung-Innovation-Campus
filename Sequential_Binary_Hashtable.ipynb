```python
egg_dic={}
for i in range(1000):
  egg_dic[i]=random.randint(0,10)
```

```python
batch_num=int(input("Enter batch num: "))
%time
def egg_break(edic, bnum):
  for key,val in edic.items():
    if key==bnum:
      print(f"Floor to break on is: {val}")
egg_break(egg_dic,batch_num)
```

```python
%time
def binary_egg_breaker(edic,bnum):

  low=0
  high=len(edic)-1

  while low<=high:

      mid=(low + high) // 2

      if mid == bnum:
        return edic[bnum]

      elif bnum < mid:
        high = mid - 1

      else:
        low = mid + 1
  return -1

print(binary_egg_breaker(egg_dic,44))
```

```python
import random

# Choose a random integer between 0 and 9 (inclusive) to represent the critical_floor.
critical_floor = random.randint(0, 9)

# Create an empty list called egg_breaking_data.
egg_breaking_data = []

# Loop from 0 to 9 (representing each floor).
for floor in range(10):
  # If the current floor number is less than the critical_floor, append 0 (representing 'safe') to egg_breaking_data.
  if floor < critical_floor:
    egg_breaking_data.append(0) # 0 for safe
  # Otherwise (if the current floor is greater than or equal to the critical_floor), append 1 (representing 'breaking') to egg_breaking_data.
  else:
    egg_breaking_data.append(1)


print(f"Critical Floor: {critical_floor}")
print(f"Egg Breaking Data: {egg_breaking_data}")
```

```python
%time
def sequential_search(egg_data):
    for floor, status in enumerate(egg_data):
        if status == 1:
            return floor
    return -1 # Should not be reached if there's always a critical floor

critical_floor_found = sequential_search(egg_breaking_data)
print(f"The critical breaking floor found by sequential search is: {critical_floor_found}")
```

```python
%time
def binary_search(egg_data):
    low = 0
    high = len(egg_data) - 1
    critical_floor = -1

    while low <= high:
        mid = low + (high - low) // 2

        if egg_data[mid] == 1: # Egg breaks at mid floor
            critical_floor = mid # This could be the critical floor
            high = mid - 1     # Try to find a lower critical floor
        else: # egg_data[mid] == 0, egg is safe
            low = mid + 1      # Critical floor must be higher
    return critical_floor

critical_floor_found_binary = binary_search(egg_breaking_data)
print(f"The critical breaking floor found by binary search is: {critical_floor_found_binary}")
```

```python
print(f"Critical floor found by sequential search: {critical_floor_found}")
print(f"Critical floor found by binary search: {critical_floor_found_binary}")
```

#normalize any text to a num range 0-100

```python
arr=[100,3,546,44,555,200]
def normalization(arr):
  new_arr=[]
  for i in arr:
    new_arr.append(((i - min(arr))/(max(arr)-min(arr)))*100)
  return new_arr
normalization(arr)
```

```python
class MyHashTable:

  def __init__(self,size):
      self.hash_table=[]
      self.size=size
      for i in range(self.size):
          self.hash_table.append([])

  def hash(self,values):
      new_val=[]
      for ch in values:
        sum_ = 0
        sum_ += len(ch)
        new_val.append((sum_) % 10)
      return new_val

  def get(self,value):
      pass

  def put(self,value,key):
      self.hash_table[key].append(value)
```

```python
h=MyHashTable(5)
h.hash(["dssd","d","uytd","ds","ddsd"])
h.put()
```
