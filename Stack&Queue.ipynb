##STACK

```python
class Stack:

  def __init__(self):
    self.a=[]

  def push(self,x):
    self.a.append(x)


  def pop(self):
    try:
      return self.a.pop()
    except Exception as e:
      return f"Error is: {e}"

  def top(self):
    try:
      return self.a[-1]
    except Exception as e:
      return f"Error is: {e}"

  def size(self):
    return len(self.a)


  def isempty(self):
    return True if len(self.a)==0 else False

  def elems(self):
    return f"Stack: {self.a[::-1]} "

```

```python
s1=Stack()
print(s1.push(1))
print(s1.push(2))
print(s1.isempty())
print(s1.top())
print(s1.size())
print(s1.pop())
print(s1.top())
print(s1.push(3))
print(s1.push(5))
print(s1.elems())
```

```python
s2=Stack()
s2.pop()
print(s2.top())
print(s2.push("Stack is empty"))
print(s2.isempty())
print(s2.elems())
print(s2.pop())
s2.isempty()
```

##QUEUE

```python
class Queue:

  def __init__(self):
    self.a=[]

  def enqueue(self,x):
    self.a.append(x)


  def dequeue(self):
    try:
      return  self.a.pop(0)
    except Exception as e:
      return f"Error is: {e}"

  def top(self):
    try:
      return  {self.a[-1]}
    except Exception as e:
      return f"Error is: {e}"

  def size(self):
    return len(self.a)


  def isempty(self):
    return "Queue is empty" if len(self.a)==0 else f"Queue has elements "

  def elems(self):
    return f"Queue: {self.a} "

  def enqueue_list(self,lis):
    self.a.extend(lis)


```

```python
q=Queue()
q.enqueue(2)
q.enqueue(3)
q.enqueue(4)
print(q.elems())
print(q.dequeue())
print(q.elems())
```

##Josephus Circle

```python
def josephus(n,k):
  a=[i for i in range(1,n+1)]
  que=Queue()
  que.enqueue_list(a)
  i=1

  while que.size()!=1:
    if i % k == 0:
      que.dequeue()
    else:
      que.enqueue(que.dequeue())
    print(que.elems())
    i+=1

  print(f"Winner: {que.dequeue()}")
```

```python
josephus(5,2)
```

```python
def bracket_problem(string):

  st=Stack()
  # Using a dictionary for direct mapping of open to closed brackets
  bracket_map = {'(': ')', '{': '}', '[': ']'}
  open_brackets = set(bracket_map.keys())
  closed_brackets = set(bracket_map.values())

  for char in string:
    if char in open_brackets:
     st.push(char)

    elif char in closed_brackets:
      if st.isempty(): # Unmatched closing bracket
        return False

      popped_open_bracket = st.pop() # Pop the last opened bracket
      # Check if the popped opening bracket's corresponding closing bracket matches the current closing bracket
      if bracket_map.get(popped_open_bracket) != char:
        return False

  # After iterating, if the stack is empty, all brackets are matched
  return st.isempty()
```

```python
bracket_problem("[{()}]")
```
