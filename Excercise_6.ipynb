Factorial Problem

```python
def factorial(n):
  return n*factorial(n-1) if n > 1 else 1

factorial(5)
```

Sum of natural Numbers

```python
def sum_natural(n):
  return n + sum_natural(n-1) if n>1 else 1

sum_natural(5)
```

Print Numbers till n

```python
def print_numbers(n):
  if n >= 1:
    print_numbers(n-1)
    print(n)
  else:
    return 0

print_numbers(10)
```

Power qns

```python
def power(base, exp):
  if exp == 0:
    return 1
  else:
    return base * power(base, exp - 1)
power(2,3)
```

Fibonacci

```python
def fibonacci(n):
  if n == 1:
    return 1
  elif n == 0:
    return 0
  else:
    return fibonacci(n-1) + fibonacci(n-2)

fibonacci(4)
```

Palindrome Question

```python
def is_palindrome(s):
  if len(s) == 1 or not s:
    return True
  if s[0] == s[-1]:
    return is_palindrome(s[1:-1])
  else:
    return False

is_palindrome('ssss')
```

Reversing a String

```python
def reverse_string(s):
  if not s:
    return s
  else:
    return reverse_string(s[1:]) + s[0]

reverse_string("like")
```

Greatest Common Divisor

```python
def gcd(a, b):
  if b == 0:
    return a
  return gcd(b , a % b)

gcd(45,10)  # a > b
```

Tower of Hanoi Problem

```python
def tower_of_hanoi(n, source, target, auxiliary):
    if n == 1:
        print(f"Move disk 1 from {source} to {target}")
        return
    tower_of_hanoi(n - 1, source, auxiliary, target)
    print(f"Move disk {n} from {source} to {target}")
    tower_of_hanoi(n - 1, auxiliary, target, source)

# Example usage for 3 disks
print("Steps to solve Tower of Hanoi for 3 disks:")
tower_of_hanoi(3, 'A', 'C', 'B')
```
