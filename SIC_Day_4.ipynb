#WAP to find shortest string in a list of string

```python
#Working Version
slist=["abc",'ab',"a","ABD","b","c"]
min=slist[0]
for st in slist:
  if len(min)>len(st):
    min=st
for st in slist:
  if len(min)==len(st) and min!=st:
    print("Shortest string is: ",st)
print("Shortest string is: ",min)


#Better Version
slist=["abc",'ab',"a","ABD","b","c","abcd"]
i=0
min=slist[0]
minl=[]
while True:
  i+=1
  if len(min)>len(slist[i]):
    min=slist[i]
  elif len(min)==len(slist[i]) and min!=slist[i]:
    minl.append(min)
    minl.append(slist[i])
  if i==len(slist)-1:
    break
print("Shortest string is: ",set(minl))

```

#WAP to find a target in a list with minimum time complexity


```python
nlist=[1,5,3,6,7,9,4,2,8]
nlist.sort()
nlist
p1=nlist[0];p2=nlist[-1]
target=int(input("Enter target:"))
while True:
  mid=(p1+p2)//2
  if target==mid:
    print("Target found!")
    break
  elif target>mid:
    p1=mid
    print("Changed")
  elif target<mid:
    p2=mid
    print("Changed")
  else:
    print("Target not found!")
    break
```

#WAP to find 1st repeated element in list

```python
#Working
nlist=[1,3,"A",4,6,"A"]
hash=[]
found=0
for i in nlist:
  if i not in hash:
    hash.append(i)
  else:
    print("First repeated Element: ", i)
    found=1
    break
if found==0:
  print("No repeated elements")
```
