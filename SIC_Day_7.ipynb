#WAP occurance of numbers in a list

```python
#Normal
a=[1,2,3,1,1,2,3,4,3,1]
hash={}
for i in a:         #O(1)
  hash[i]=1 if i not in hash else hash[i]+1
print(hash)
```

```python
count={}
count.get(2,0)      #If val give val else 0
```

```python
#Better
for i in a:
  count[i] = count.get(i,0)+1
print(count)
```

```python
#More optimized
from collections import defaultdict
count=defaultdict(int)
for i in a:
  count[i]+=1
print(count)
```

```python
#Easiest
from collections import Counter
count=Counter(a)
print(count)
```

#WAP to get sum and diff of two numbers

```python
def sum_diff (a = 0, b = 0):
  return a + b, a - b

print(sum_diff(3,7))

def sum_diff (a = 0, b = 0):
  return f"Sum: {a + b}, Difference : {a - b} of {a}, {b}"

print(sum_diff(3,7))
print(sum_diff(3,b=5))
```
