```python
n=0
def check(num,arr):
  sum=num
  if sum == 0:
    return True

  if num % arr[0] == 0:
    return True

  if sum < 0:
    global n
    n+=1
    try:
      return check(sum,arr[n::])
    except :
      return False

  sum=num
  sum-=arr[0]
  return check(sum,arr[1::])

```

```python
l=[2,5]
n=4
check(n,l)
```

```python
def best_sum_memo(target_sum, numbers, memo=None):
    if memo is None:
        memo = {}

    if target_sum in memo:
        return memo[target_sum]

    if target_sum == 0:
        return []

    if target_sum < 0:
        return None
```

```python
def best_sum_memo(target_sum, numbers, memo=None):
    if memo is None:
        memo = {}

    if target_sum in memo:
        return memo[target_sum]

    if target_sum == 0:
        return []

    if target_sum < 0:
        return None

    shortest_combination = None

    for num in numbers:
        remainder = target_sum - num
        remainder_combination = best_sum_memo(remainder, numbers, memo)
        if remainder_combination is not None:
            combination = remainder_combination + [num]
            if shortest_combination is None or len(combination) < len(shortest_combination):
                shortest_combination = combination

    memo[target_sum] = shortest_combination
    return shortest_combination
```

```python
l=[1,2,5]
n=7
print(best_sum_memo(n, l))
```

```python
def best_sum(n,l):
  if n==0:
    return []
  if n< 0:
    return None
  answer = []
  for i in l:
    res=best_sum(n-i, l)
    if res is not None:
      res.append(i)
      answer.append(res)
  return min(answer, key=len, default=None)
```

```python
best_sum(40, [10,5,25,20])
```

#Bottom up approach

```python
def fib(N):
  table=[0]*(N+1)
  table[0]=1;table[1]=1

  for i in range(2,N+1):
    table[i]=table[i - 1] + table[i - 2]

  return table[N]
```

```python
fib(5)
```
