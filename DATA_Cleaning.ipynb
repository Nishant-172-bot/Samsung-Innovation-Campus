```python
age_df=df.dropna(axis=0, how='any',subset=['age'])    #delete row if any missing value -> how; subset of columns to consider
len(age_df)
```

```python
age_df
```

```python
#replacing data:
#Replacing with mean -> standard data, median -> if outliers, mode->categorical data, nearest neighbour -> time series data(temp)
avg_age=df['age'].mean(axis=0)
avg_age
```

```python
median_age=df['age'].median(axis=0)
median_age
```

```python
town_count=df['embark_town'].value_counts(dropna=True)
town_count
```

```python
most=town_count.idxmax()
most
```

```python
df['embark_town']=df['embark_town'].fillna(most) #df.isfillna({'embark_town:most},inplace=True)
df.isnull().sum()
```

```python
df_01=df['embark_town'].ffill() #Mathi ko value filled to NaN -> frontfill(ffill) and another bakfill(bfill)
df_01
```
