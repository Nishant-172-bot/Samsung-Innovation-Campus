```python
close=df["Close"]["AAPL"]
ma5=df[("Close","AAPL")].rolling(window=5).mean()
ma10=df[("Close","AAPL")].rolling(window=10).mean()
ma60=df[("Close","AAPL")].rolling(window=60).mean()
ma120=df[("Close","AAPL")].rolling(window=120).mean()
plt.figure(figsize=(15,10))
plt.plot(close,label='close',linewidth=5)
plt.plot(ma5,label='ma5')
plt.plot(ma10,label='ma10')
plt.plot(ma60,label='ma60')
plt.plot(ma120,label='ma120')
plt.legend()
plt.title("Rolling Window Calculation")
plt.xlabel("Date")
plt.ylabel("Closing Price")
```

```python
fs=pd.read_csv("/content/capture-fisheries-vs-aquaculture.csv")
fs
```

```python
fs.info()
```

```python
fs.drop(['Code'], axis=1, inplace=True)
fs
```

```python
fs.isnull().sum()
```

```python
fs.fillna(0,inplace=True)
fs.isnull().sum()
```

```python
fs["New_Year"]=pd.to_datetime(fs['Year'].astype(str) , format="%Y")
fs.set_index('New_Year',inplace=True)
fs.drop(['Year'], axis=1, inplace=True)
```

```python
fs
```

```python
new_df=fs.sort_index()
```

```python
new_df
```

```python
new_df['Entity']=new_df['Entity'].astype("category")
new_df
```

```python
g=new_df.groupby(["New_Year"])    #shows only data for one date -> compress
world_total=g.sum(numeric_only=True)
world_total.head()      #sum year wise not country wise
```

```python
g.head(1)
```

```python
plt.style.use('ggplot')
world_total.plot(kind='area', alpha=0.2, stacked=False, figsize=(20,10))
plt.legend()
plt.show()
```
