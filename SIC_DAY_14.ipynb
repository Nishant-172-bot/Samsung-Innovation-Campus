##HashTable

```python
class MyHashTable:

    def __init__(self, size):
        self.size = size
        self.hash_table = [[] for _ in range(size)]

    def hash(self, key):
        hash_sum = 0
        for ch in str(key):
            hash_sum += ord(ch)
        return hash_sum % self.size

    def put(self, key, value=0):
        index = self.hash(key)

        for i, (k, v) in enumerate(self.hash_table[index]):
            if k == key :
                self.hash_table[index][i] = (key, value)
                return

        self.hash_table[index].append((key, value))

    def get(self, key):
        index = self.hash(key)

        for k, v in self.hash_table[index]:
            if k == key:
                return v

        return None
```

```python
ht = MyHashTable(10)

ht.put("name", "Alice")
ht.put("age", 25)
ht.put("age",66)

print(ht.get("name"))
print(ht.get("age"))
print(ht.get("job"))
print(ht.hash_table)
```
