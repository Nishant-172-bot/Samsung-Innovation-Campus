```python
print("1)Addition   2)Subtraction   3)Multiplication   4)Division   :")
ch=int(input("Enter choice:"))
num1=int(input("Enter two numbers for operation"))
num2=int(input())
if ch==1:
  print(f"{num1} + {num2} = ",num1 + num2)
elif ch==2:
  print(f"{num1} - {num2} = ",num1 - num2)
elif ch==3:
  print(f"{num1} * {num2} = ",num1 * num2)
elif ch==2:
  print(f"{num1} / {num2} = ",num1 / num2)
else:
  print("Incorrect Choice")
```

```python
#WAP to give quadrant from two coordinates
coord =input("Give two coords ")
clist=coord.split(sep="/")

if int(clist[0]) == 0 :
  print("On Y-Axis line")
elif int(clist[1]) == 0:
  print("On X-Axis line")
elif int(clist[0]) == 0 and int(clist[1]) == 0:
  print("Origin")
else:
  if int(clist[0]) >= 0 and int(clist[1]) >= 0:
    print("First Quadrant")
  elif int(clist[0]) <= 0 and int(clist[1]) >= 0:
    print("Second Quadrant")
  elif int(clist[0]) <= 0 and int(clist[1]) <= 0:
    print("Third Quadrant")
  elif int(clist[0]) >= 0 and int(clist[1]) <= 0:
    print("Fourth Quadrant")
```

```python
start = 1

for i in range(5):
  if i % 2 == 0:
    for j in range(start , start + 5):
      print(j, end="\t")
  else:
    for j in range(start + 4 , start  - 1 , -1):
      print(j, end="\t")
  start += 5
  print()
```

```python
import random
count=0
num=random.randint(1,100)
while True:
  count+=1
  guess=int(input("Enter a Number: "))
  if num<guess:
    print("Too High")
  elif num>guess:
    print("Too Low")
  elif num==guess:
    print("Congrats! Total Try= ",count)
    break
  else:
    continue
```
