```python
def anagram(s1,s2):
    if len(s1) != len(s2):
        return False
    count = {}

    for ch in s1:
        count[ch] = count.get(ch, 0) + 1

    for ch in s2:
        if ch not in count:
            return False
        count[ch] -= 1
        if count[ch] < 0:
            return False

    return True
anagram('silent','listen')
```

```python
def anagram_sorted(s1, s2):
    if len(s1) != len(s2):
        return False

    return sorted(s1) == sorted(s2)

print(anagram_sorted("listen", "silent"))
print(anagram_sorted("hello", "world"))
```

```python
def selection_sort2(arr):
    n = len(arr)
    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr
```

```python
def anagram_selection(s1, s2):
    if len(s1) != len(s2):
        return False

    list1 = list(s1)
    list2 = list(s2)
    selection_sort2(list1)
    selection_sort2(list2)

    return list1 == list2

print(anagram_selection("listen", "silent"))
```

```python
def insertion_sort2(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key
    return arr
```

```python
def anagram_insertion(s1, s2):
    if len(s1) != len(s2):
        return False

    list1 = list(s1)
    list2 = list(s2)
    insertion_sort2(list1)
    insertion_sort2(list2)

    return list1 == list2

print(anagram_insertion("listen", "silent"))
```

```python
def twoway_merge(l1, l2):
    result = []
    i = j = 0

    while i < len(l1) and j < len(l2):
        if l1[i] < l2[j]:
            result.append(l1[i])
            i += 1
        else:
            result.append(l2[j])
            j += 1

    result.extend(l1[i:])
    result.extend(l2[j:])

    return result


def multiway_merge(lists):
    if not lists:
        return []

    while len(lists) > 1:
        merged = []

        for i in range(0, len(lists), 2):
            l1 = lists[i]
            l2 = lists[i+1] if i+1 < len(lists) else []
            merged.append(twoway_merge(l1, l2))

        lists = merged

    return lists[0]
```

```python
# Given the list below, write the output after executing the partition1() function.
# S=[15,10,12,20,25,13,22]
# partition1(S,0,len(S)-1)
# print(S)
#Answer: [15,10,12,20,13,22,25]
```

```python
%time
def partition_first(arr, low, high):
    pivot = arr[low]
    left = low + 1
    right = high

    while True:
        while left <= right and arr[left] <= pivot:
            left += 1
        while left <= right and arr[right] >= pivot:
            right -= 1
        if left > right:
            break
        arr[left], arr[right] = arr[right], arr[left]

    arr[low], arr[right] = arr[right], arr[low]
    return right


def quicksort_first(arr, low, high, k):
    if low <= high:
        pivot_index = partition_first(arr, low, high)

        if pivot_index == k:
            return arr[pivot_index]
        elif pivot_index > k:
            return quicksort_first(arr, low, pivot_index - 1, k)
        else:
            return quicksort_first(arr, pivot_index + 1, high, k)


def kth_largest_first(arr, k):
    n = len(arr)
    return quicksort_first(arr, 0, n-1, n-k)

arr = [7,10,4,3,20,15]
print(kth_largest_first(arr.copy(), 3))   # 10
```

```python
%time
import random

def partition_random(arr, low, high):
    rand_index = random.randint(low, high)
    arr[rand_index], arr[high] = arr[high], arr[rand_index]

    pivot = arr[high]
    i = low

    for j in range(low, high):
        if arr[j] <= pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1

    arr[i], arr[high] = arr[high], arr[i]
    return i


def quicksort_random(arr, low, high, k):
    if low <= high:
        pivot_index = partition_random(arr, low, high)

        if pivot_index == k:
            return arr[pivot_index]
        elif pivot_index > k:
            return quicksort_random(arr, low, pivot_index - 1, k)
        else:
            return quicksort_random(arr, pivot_index + 1, high, k)


def kth_largest_random(arr, k):
    n = len(arr)
    return quicksort_random(arr, 0, n-1, n-k)

arr = [7,10,4,3,20,15]
print(kth_largest_random(arr.copy(), 3))   # 10
```
